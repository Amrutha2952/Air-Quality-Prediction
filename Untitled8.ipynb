```python
# Data Manipulation
import pandas as pd
import numpy as np

# Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning & Preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Forecasting
from prophet import Prophet

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
file_path = "/content/air_pollution_data.csv"

try:
    df = pd.read_csv(file_path)
except:
    df = pd.read_csv(file_path, engine="python")

df.head()
```

```python
print("Shape:", df.shape)
df.head()
df.info()
df.describe()
df.isnull().sum()
```

```python
#Replace invalid values (-200) with NaN
df.replace(-200, np.nan, inplace=True)

#Select only numeric columns
numeric_cols = df.select_dtypes(include=np.number).columns

#Fill missing values in numeric columns with column mean
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

#Verify missing values are handled
df.isnull().sum()
```

```python
#Convert date column to datetime
df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y')

#Check conversion
df.head()
```

```python
#Select relevant features
features = ['co', 'no', 'no2', 'o3', 'so2', 'pm2_5', 'pm10', 'nh3', 'aqi']
data = df[features]

#Scale numeric features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

scaled_df = pd.DataFrame(scaled_data, columns=features)
scaled_df['date'] = df['date']
scaled_df.head()
```

```python
# Prophet requires columns: 'ds' (datetime) and 'y' (target)
prophet_df = df[['date', 'aqi']].rename(columns={'date':'ds', 'aqi':'y'})
prophet_df.head()
```

```python
# Split into training and testing sets (80-20 split)
train_size = int(len(prophet_df) * 0.8)
train_df = prophet_df[:train_size]
test_df = prophet_df[train_size:]

train_df.tail(), test_df.head()
```

```python
# Initialize Prophet model
model = Prophet(daily_seasonality=True)

# Train model
model.fit(train_df)
```

```python
# Create dataframe for future predictions (matching test set length)
future = model.make_future_dataframe(periods=len(test_df))

# Make predictions
forecast = model.predict(future)

# Rename columns
forecast = forecast.rename(columns={
    'ds': 'Date',
    'yhat': 'Predicted_AQI',
    'yhat_lower': 'Lower_Bound_AQI',
    'yhat_upper': 'Upper_Bound_AQI'
})

# Display last few rows
forecast[['Date', 'Predicted_AQI', 'Lower_Bound_AQI', 'Upper_Bound_AQI']].tail()

```

```python
# Extract predictions corresponding to test set
predicted = forecast.set_index('Date').loc[test_df['ds']]
y_true = test_df['y'].values
y_pred = predicted['Predicted_AQI'].values

# Compute metrics
mae = mean_absolute_error(y_true, y_pred)
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
r2 = r2_score(y_true, y_pred)

print(f"Mean Absolute Error: {mae:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
print(f"R-squared: {r2:.2f}")
```

```python
plt.figure(figsize=(12,6))

# Plot actual AQI
plt.plot(test_df['ds'], y_true, color='blue', marker='o', linestyle='None', label='Actual AQI')

# Plot predicted AQI
plt.plot(test_df['ds'], y_pred, color='orange', label='Predicted AQI')

# Plot uncertainty band
plt.fill_between(test_df['ds'], predicted['Lower_Bound_AQI'], predicted['Upper_Bound_AQI'],
                 color='orange', alpha=0.2, label='Prediction Interval')

plt.xlabel('Date')
plt.ylabel('AQI')
plt.title('Air Quality Prediction with Uncertainty')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Create dataframe for next 7 days
future_7days = model.make_future_dataframe(periods=7)

# Predict future AQI
forecast_7days = model.predict(future_7days)

# Rename columns for clarity
forecast_7days = forecast_7days.rename(columns={
    'ds': 'Date',
    'yhat': 'Predicted_AQI',
    'yhat_lower': 'Lower_Bound_AQI',
    'yhat_upper': 'Upper_Bound_AQI'
})

# Display predictions for the next 7 days
forecast_7days[['Date', 'Predicted_AQI', 'Lower_Bound_AQI', 'Upper_Bound_AQI']].tail(7)
```

```python
plt.figure(figsize=(10,5))

# Plot predicted AQI
plt.plot(forecast_7days['Date'][-7:], forecast_7days['Predicted_AQI'][-7:], color='orange', marker='o', label='Predicted AQI')

# Plot uncertainty band
plt.fill_between(forecast_7days['Date'][-7:],
                 forecast_7days['Lower_Bound_AQI'][-7:],
                 forecast_7days['Upper_Bound_AQI'][-7:],
                 color='orange', alpha=0.2, label='Prediction Interval')

plt.xlabel('Date')
plt.ylabel('AQI')
plt.title('Air Quality Forecast for Next 7 Days')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
